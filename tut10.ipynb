

Les dictionnaires sont une structure de données non séquentielle, où chaque élément est une paire clé-valeur, et l'accès se fait via des clés uniques plutôt qu'un index.
Les dictionnaires sont muables, c'est à dire que sa valeur peut être modifiée.


Une clé est majoritairement représenté par deux types de données : des chaînes de caractères ou des valeurs numériques. *
* En réalité, une clé peut être tout type de données immuables (valeur booléenne, tuple, nombre, chaîne, etc.)


```python
dict1 = {
    "giraffe": 10,
    "singe": {
        "gorille": 2,
        "chimpanzee": 10
    }
}
print(dict1)
```

```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches

fig, ax = plt.subplots()
ax.set_aspect('equal')

# Liste de rectangles (x, y, w, h, couleur, épaisseur contour)
rectangles = [
    (5, 5, 20, 10, 'skyblue', 2),
    (30, 10, 15, 25, 'lightgreen', 3),
    (50, 15, 40, 15, 'orange', 1),
    (10, 30, 25, 20, 'purple', 2),
]

for (x, y, w, h, color, lw) in rectangles:
    rect = patches.Rectangle((x, y), w, h,
                             linewidth=lw, edgecolor='black', facecolor=color, alpha=0.6)
    ax.add_patch(rect)

ax.set_xlim(0, 100)
ax.set_ylim(0, 60)
plt.title("Exemple de plusieurs rectangles")
plt.grid(True)
plt.show()
```

```python
# Définir les coordonnées des sommets du triangle
triangle = patches.Polygon([[2, 2], [4, 6], [6, 2]], closed=True, facecolor='tomato', edgecolor='black')

# Créer un cercle
circle = patches.Circle((4, 4), radius=2, facecolor='lightgreen', edgecolor='black')

# --- Parallélogramme ---
parallelogramme = patches.Polygon([[1, 1], [4, 1], [5, 3], [2, 3]], closed=True, facecolor='lightblue', edgecolor='black')
ax.add_patch(parallelogramme)

# --- Triangle ---
triangle = patches.Polygon([[6, 1], [8, 4], [10, 1]], closed=True, facecolor='lightgreen', edgecolor='black')
ax.add_patch(triangle)

# --- Cercle ---
circle = patches.Circle((3, 5), radius=2, facecolor='lightcoral', edgecolor='black')
ax.add_patch(circle)

# --- Trapèze ---
trapeze = patches.Polygon([[6, 5], [9, 5], [8, 7], [7, 7]], closed=True, facecolor='lightyellow', edgecolor='black')
ax.add_patch(trapeze)
```

```python

import matplotlib.pyplot as plt

# Données de la chute libre
x = [0, 1, 2, 3]
y = [0, 9.8, 29.4, 58.8]
# Données de la vitesse constante (5 m/s)
y2 = [0, 5, 10, 15]
# Données de la vitesse nulle (position constante de 2 m)
y3 = [2, 2, 2, 2]

# Tracer les trois courbes
plt.plot(x, y, color="red", marker="o", label="Chute libre")
plt.plot(x, y2, color="blue", marker="o", label="Vitesse constante")
plt.plot(x, y3, color="green", marker="o", label="Vitesse nulle")

# Ajouter le titre et les noms des axes
plt.title("Déplacement d'un objet dans le temps")
plt.xlabel("Temps écoulé (s)")
plt.ylabel("Position de l'objet (m)")

# Ajouter la légende
plt.legend()

# Afficher le graphique
plt.show()
```

```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches

# Création de la figure et des axes
fig, ax = plt.subplots()

# Ajout d'un cercle (original)
original_circle = patches.Circle((0.3, 0.5), radius=0.1, color='blue', alpha=0.6, label='Cercle Original')
ax.add_patch(original_circle)

# Ajout du cercle symétrique par rapport à l'axe vertical (x=0)
symmetric_circle = patches.Circle((-0.3, 0.5), radius=0.1, color='red', alpha=0.6, label='Cercle Symétrique')
ax.add_patch(symmetric_circle)

# Définition des limites du graphique
ax.set_xlim(-1, 1)
ax.set_ylim(0, 1)

# Ajout des labels et de la grille
ax.set_title("Symétrie du Cercle")
ax.set_xlabel("Axe X")
ax.set_ylabel("Axe Y")
ax.grid(True)

# Affichage de la légende
ax.legend()

# Affichage du graphique
plt.show()
```
