

Les dictionnaires sont une structure de données non séquentielle, où chaque élément est une paire clé-valeur, et l'accès se fait via des clés uniques plutôt qu'un index.
Les dictionnaires sont muables, c'est à dire que sa valeur peut être modifiée.


Une clé est majoritairement représenté par deux types de données : des chaînes de caractères ou des valeurs numériques. *
* En réalité, une clé peut être tout type de données immuables (valeur booléenne, tuple, nombre, chaîne, etc.)


```python
dict1 = {
    "giraffe": 10,
    "singe": {
        "gorille": 2,
        "chimpanzee": 10
    }
}
print(dict1)
```

```python
def total_ventes(ventes):
    somme = 0
    for produit in ventes:
        somme += ventes[produit]
    return somme

def ventes_adjustees(ventes):
    moy = total_ventes(ventes) / len(ventes)  # Calcul de la moyenne des ventes
    ventes_modifiees = ventes.copy()  # Faire une copie pour ne pas modifier le dictionnaire original
    for produit, quantite in ventes.items():
        if quantite < moy:
            ventes_modifiees[produit] = moy  # Remplacer par la moyenne
    return ventes_modifiees

# Dictionnaire des ventes des produits
ventes_magasin = {
    "Pommes": 150,
    "Bananes": 80,
    "Oranges": 200,
    "Poires": 60,
    "Raisins": 120
}

# Affichage des ventes ajustées
print(ventes_adjustees(ventes_magasin))
```

```python
# Ajoute une évaluation pour un patient
# patient_scores : Registre des scores des patients
# patient : Nom du patient à évaluer
# score : Score de l'évaluation à ajouter
#
# Retourne le dictionnaire mis à jour
def ajouter_evaluation(patient_scores, patient, score):
    if patient in patient_scores:
        patient_scores[patient].append(score)
    else:
        patient_scores[patient] = [score]
    return patient_scores

# Calcule la moyenne des scores d'un patient
# patient_scores : Registre des scores des patients
# patient : Nom du patient pour lequel calculer la moyenne
#
# Retourne la moyenne des scores ou un message si le patient n'est pas enregistré
def calculer_moyenne(patient_scores, patient):
    if patient not in patient_scores:
        return f"Le patient {patient} n'est pas enregistré."

    scores = patient_scores[patient]
    return sum(scores) / len(scores)

# Exemple d'utilisation
patient_scores = {
    "Alice": [85, 90, 78],
    "Bob": [70, 75, 80]
}

ajouter_evaluation(patient_scores, "Alice", 92)
ajouter_evaluation(patient_scores, "Charlie", 88)
print(calculer_moyenne(patient_scores, "Bob"))
print(calculer_moyenne(patient_scores, "Charlie"))
```

```python

import matplotlib.pyplot as plt

# Données de la chute libre
x = [0, 1, 2, 3]
y = [0, 9.8, 29.4, 58.8]
# Données de la vitesse constante (5 m/s)
y2 = [0, 5, 10, 15]
# Données de la vitesse nulle (position constante de 2 m)
y3 = [2, 2, 2, 2]

# Tracer les trois courbes
plt.plot(x, y, color="red", marker="o", label="Chute libre")
plt.plot(x, y2, color="blue", marker="o", label="Vitesse constante")
plt.plot(x, y3, color="green", marker="o", label="Vitesse nulle")

# Ajouter le titre et les noms des axes
plt.title("Déplacement d'un objet dans le temps")
plt.xlabel("Temps écoulé (s)")
plt.ylabel("Position de l'objet (m)")

# Ajouter la légende
plt.legend()

# Afficher le graphique
plt.show()
```

```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches

# Création de la figure et des axes
fig, ax = plt.subplots()

# Ajout d'un cercle (original)
original_circle = patches.Circle((0.3, 0.5), radius=0.1, color='blue', alpha=0.6, label='Cercle Original')
ax.add_patch(original_circle)

# Ajout du cercle symétrique par rapport à l'axe vertical (x=0)
symmetric_circle = patches.Circle((-0.3, 0.5), radius=0.1, color='red', alpha=0.6, label='Cercle Symétrique')
ax.add_patch(symmetric_circle)

# Définition des limites du graphique
ax.set_xlim(-1, 1)
ax.set_ylim(0, 1)

# Ajout des labels et de la grille
ax.set_title("Symétrie du Cercle")
ax.set_xlabel("Axe X")
ax.set_ylabel("Axe Y")
ax.grid(True)

# Affichage de la légende
ax.legend()

# Affichage du graphique
plt.show()
```
