## Boucle:
while : (lorsque nous devons répéter une opération plusieurs fois, le nombre d'itérations à réaliser pour cette boucle n'est pas connu à l'avance)

for : (nous savons à l'avance combien d'itération il faudra effectuer. )

break : (le programme sortira de la boucle)

continue : (le programme passe à la prochaine itération de la boucle sans compléter l'itération en cours)

```python
# Demande à l'utilisateur de saisir la base
base = float(input("Entrez la base : "))

# Demande à l'utilisateur de saisir l'exposant
exposant = int(input("Entrez l'exposant : "))

# Initialisation de la variable result qui va stocker le résultat
resultat = 1

# Boucle pour multiplier la base par elle-même
for _ in range(abs(exposant)):
    resultat *= base

# Si l'exposant est négatif, on inverse le résultat
if exposant < 0:
    resultat = 1 / resultat

# Affichage du résultat
print(f"{base} élevé à la puissance {exposant} est {resultat}")
```

# q6


```python
valide = False
while not valide:
    # Demande de saisie du mot de passe
    mot_de_passe = input("Veuillez entrer votre mot de passe : ")

    # Quitter le programme si on écrit stop
    if mot_de_passe.lower() == 'stop':
        break

    # Vérifie la longueur
    if len(mot_de_passe) < 8:
        print("Le mot de passe doit contenir au moins 8 caractères.")
        continue

    contient_majuscule = False
    contient_minuscule = False
    contient_chiffre = False
    contient_special = False

    # Les caractères spéciaux admis
    caracteres_speciaux = "!@#$%^&*()-_=+[]{}|;:'\",.<>?/`~"

    for caractere in mot_de_passe:
        if caractere.isupper():
            contient_majuscule = True
        elif caractere.islower():
            contient_minuscule = True
        elif caractere.isdigit():
            contient_chiffre = True
        elif caractere in caracteres_speciaux:
            contient_special = True

    # Autres tests
    if not contient_majuscule:
        print("Le mot de passe doit contenir au moins une lettre majuscule.")
        continue
    if not contient_minuscule:
        print("Le mot de passe doit contenir au moins une lettre minuscule.")
        continue
    if not contient_chiffre:
        print("Le mot de passe doit contenir au moins un chiffre.")
        continue
    if not contient_special:
        print("Le mot de passe doit contenir au moins un caractère spécial.")
        continue

    # Tous les tests ont passés
    print("Le mot de passe est valide.")
    valide = True

```

# q8


```python
n = int(input("Entrez le nombre de premiers nombres premiers à afficher : "))

compte = 0  # Compteur de nombres premiers trouvés
nombre = 2   # Le premier nombre à tester
while compte < n:
    est_premier = True
    for i in range(2, int(nombre**0.5) + 1):
        if nombre % i == 0:
            est_premier = False

    if est_premier:
        print(nombre)  # Affiche le nombre premier
        compte += 1    # Incrémente le compteur

    nombre += 1  # Teste le nombre suivant

```

```python
import math

n = 10
x = math.pi/2

# Complétez le code pour calculer s
s = 0
for k in range(n+1):
    s = s + (-1)**k * (pow(x, 2*k + 1)) / math.factorial(2*k + 1)

print("Résultat pour n =", n, "et x =", x, ": s =", s)

# Résultat attendu pour n = 10 et x = pi/2 : s = 1.000000000
```
