## 6.Boucle:
while : (lorsque nous devons répéter une opération plusieurs fois, le nombre d'itérations à réaliser pour cette boucle n'est pas connu à l'avance)

for : (nous savons à l'avance combien d'itération il faudra effectuer. )

break : (le programme sortira de la boucle)

continue : (le programme passe à la prochaine itération de la boucle sans compléter l'itération en cours)



```python
'''
while <condition>:
    opération1
    opération2
    ...

for i in range(n) :
    opération1
    opération2
    ...



s = 0
for i in range(2, 10, 2):    #2,4,6
    if i == 6:
        break
    s += 1                   #1+1

print(s)                     #2



s = 0
for i in range(5, 10):       #5,6,7,8,9
    if i >= 7:
        continue
    s += i                   #0+5+6

print(s)                     #11

'''
```

## 7.procédures et fonctions:
Une fonction est une portion de code pouvant être exécutée à plusieurs reprises avec différents paramètres. (Avec ou sans paramètres)

les fonctions ont une particularité spéciale, elles retournent des valeurs(mot clé: return). Alors que les procédures ne retournent pas de valeurs.

```python
#procédure:
def bonjour(name):
    print("Bonjour " + name + "!")
    print("Bienvenue!")

name = 'Jia'
bonjour(name)


#fonction:
def bonjour(name):
    print("Bonjour " + name + "!")
    return("Bienvenue!")

name = 'Jia'
bonjour(name)
```

## 8.Modules et Tests unitaires
Module: Un module en Python est un fichier .py contenant du code Python (fonctions, variables, classes, etc.).
Il existe plusieurs modules prédéfinies dans Python qui sont intégrés de base que vous avez déjà utilisé. Notamment math, random et time.



Un test unitaire: est un type de test automatisé qui vérifie qu'une unité spécifique de code fonctionne correctement. L'unité peut être une fonction, une méthode, ou une petite portion de programme.






```python
#Modules:
import math
from math import sqrt #specifique fonction du module math




#Tests unitaires:
# Fonction à tester
def addition(a, b):
    return a + b

# Fonction pour simuler un test unitaire
def test_addition():
    # Cas 1 : 2 + 3 = 5
    if addition(2, 3) != 5:
        print("Test 1 échoué : addition(2, 3) != 5")

    # Cas 2 : -1 + 1 = 0
    if addition(-1, 1) != 0:
        print("Test 2 échoué : addition(-1, 1) != 0")

    # Cas 3 : 0 + 0 = 0
    if addition(0, 0) != 0:
        print("Test 3 échoué : addition(0, 0) != 0")

# Appeler la fonction de test
test_addition()
```
