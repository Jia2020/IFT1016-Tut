# Normalisation du code d’un acide aminé


```python
phrase = input("Entrez une phrase contenant un code d’acide aminé : ")

# Trouver le mot de 3 lettres (on suppose qu'il est là)
position_code = phrase.find(" ")#7
#print(position_code)
code = phrase[position_code+1:position_code+4]  # On isole les 3 lettres du code
code_formate = code.capitalize()

phrase_corrigee = phrase.replace(code, code_formate)

print("Phrase corrigée :", phrase_corrigee)

```

# Extraction et correction automatique



```python
# Demande de l'équation
equation = input("Entrez une équation de vitesse : ")

# Nettoyage : minuscules, suppression des espaces
equation = equation.lower().replace(" ", "")

# Séparation en deux parties
parties = equation.split("=")  # ["v", "d/t"]
gauche = parties[0].strip()
#droite = parties[1].strip().replace("/", "  /")
d=parties[1].split('/')[0]
t = parties[1].split('/')[1]

# Reconstruction propre
equation_corrigee = gauche + " = " + d + ' / '+ t

print("Équation corrigée :", equation_corrigee)

# Extraction des grandeurs physiques
#variables = [gauche] + droite.replace(" ", "").split("/")
variables = [gauche]+[d]+[t]
print(variables)# ['v','d','t']

print("Variables détectées :")
print("-", variables[0], ": vitesse")
print("-", variables[1], ": distance")
print("-", variables[2], ": temps")
```

# Les atomes d'une formule chimique



```python
# Entrée de l'utilisateur
formule = input("Entrez la formule chimique à analyser : ")

# Extraction des réactifs et produits
reactifs = formule[0:formule.find("->")]
produits = formule[formule.find("->")+2:]

# Extraction des éléments pertinents des réactifs
o2_reactifs = reactifs.count("O2")
o_reactifs = reactifs.count("O")
deux_h2_reactifs = reactifs.count("2H2")
deux_h20_reactifs = reactifs.count("2H2O")
h_reactifs = reactifs.count("H")
cl_reactifs = reactifs.count("Cl")
h2_reactifs = reactifs.count("+H2")
na_reactifs = reactifs.count("Na")
print(o2_reactifs,o_reactifs,deux_h2_reactifs,deux_h20_reactifs,h_reactifs,cl_reactifs,h2_reactifs,na_reactifs)

# Comptage de chaque élément selon les extractions des réactifs
h_bilan_reactifs = h_reactifs + 3 * deux_h2_reactifs + h2_reactifs
cl_bilan_reactifs = cl_reactifs
o_bilan_reactifs = o_reactifs + o2_reactifs + deux_h20_reactifs
na_bilan_reactifs = na_reactifs

# Extraction des éléments pertinents des produits
o2_produits = produits.count("O2")
deux_h2_produits = produits.count("2H2")
o_produits = produits.count("O")
deux_h20_produits = produits.count("2H2O")
h_produits = produits.count("H")
cl_produits = produits.count("Cl")
h2_produits = produits.count("+H2")
na_produits = produits.count("Na")

# Comptage de chaque élément selon les extractions des réactifs
h_bilan_produits = h_produits + 3 * deux_h2_produits + h2_produits
cl_bilan_produits = cl_produits
o_bilan_produits = o_produits + o2_produits + deux_h20_produits
na_bilan_produits = na_produits

print("BILAN DE LA RÉACTION")
print("Atomes d'oxygène (O) :", o_bilan_produits, "contre", o_bilan_reactifs)
print("Atomes d'hydrogène (H) :", h_bilan_produits, "contre", h_bilan_reactifs)
print("Atomes de chlore (Cl) :", cl_bilan_produits, "contre", cl_bilan_reactifs)
print("Atomes d'oxygène (Na) :", na_bilan_produits, "contre", na_bilan_reactifs)
```
