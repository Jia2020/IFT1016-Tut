# Distance entre deux points



```python
import math
# Coordonnées du point 1
x1 = 11
y1 = 15

# Coordonnées du point 2
x2 = 22
y2 = 34

# Calcul de la distance
distance = math.sqrt((x2-x1)**2+(y2-y1)**2)
round(distance,2)
```

# Notes finales



```python
# Notes dans chaque matière
note_informatique = 88
note_francais = 80
note_mathematiques = 72

# Calcul de la moyenne
moyenne = (note_informatique + note_francais + note_mathematiques) / 3

# Affichage
print("La moyenne de l'étudiant est de", round(moyenne, 2), '%')


# ----- BONUS ----- #
moyenne_ponderee = (
                    note_informatique * 0.35
                    + note_francais * 0.4
                    + note_mathematiques * 0.25
                )

# Affichage
print("La moyenne pondérée l'étudiant est de", round(moyenne_ponderee, 2), '%')
```

# Superficie d'une pyramide



```python
import math
# Déclaration de nos variables
a = 10
h = 15

# Calculs
h_lat = math.sqrt((a/2)**2 + h**2)
air_triangle = a * h_lat/2
air_carre = a*a
air_pyramide = air_carre + 4*air_triangle

# Affichage
print("La superficie de la pyramide est: ", round(air_pyramide,2),'cm^2')
```
