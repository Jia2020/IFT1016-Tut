# Calcul de la température moyenne


```python
matin = float(input("Entrez la température du matin: "))
midi = float(input("Entrez la température du midi: "))
soir = float(input("Entrez la température du soir: "))

moyenne = (matin + midi + soir)/3
print("Température moyenne de la journée :", round(moyenne,2))
```

# Une particule en MRUA



```python

x0 = float(input("Entrez la position initiale en mètres :"))
v0 = float(input("Entrez la vitesse initiale en mètres par seconde :"))
a = float(input("Entrez l'accélération de la particule en mètres par seconde carrée :"))
t = float(input("Enfin, entrez le temps de fin désiré en secondes :"))

x_t = x0 + v0 * t + 1/2 * a * t**2
v_t = v0 + a * t
distance = x_t - x0
v_moyenne = distance/t

print("Position finale :",round(x_t,2),"m")
print("Vitesse finale :",round(v_t,2),"m/s")
print("distance parcourue:",distance,"m")
print("vitesse moyenne:",v_moyenne,"m/s")
```
